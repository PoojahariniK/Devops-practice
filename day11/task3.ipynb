```python
import pandas as pd
import numpy as np

df = pd.read_csv("ecommerce_small.csv")

print("First 5 rows:")
print(df.head())

print("\nDataset shape:", df.shape)

print("\nDataset information:")
df.info()

print("\nDescriptive statistics:")
print(df.describe(include="all"))
```

```python
print("Missing values before cleaning:")
print(df.isnull().sum())

print("\nMissing value percentage:")
print((df.isnull().sum() / len(df) * 100).round(2))
```

```python
df = df.dropna().copy()

print("\nMissing values after removing rows:")
print(df.isnull().sum())
```

```python
print("Repeated OrderIDs:", df["OrderID"].duplicated().sum())
print("Exact duplicate rows:", df.duplicated().sum())
```

```python
duplicate_columns = [
    "Product", "Quantity", "UnitPrice", "City", "Customer"
]
print("Repeated transaction details:")
print(
    df[df.duplicated(subset=duplicate_columns, keep=False)]
)
```

```python
df["Customer"] = df["Customer"].str.strip().str.casefold()
df["Product"] = df["Product"].str.strip().str.title()
df["City"] = df["City"].str.strip().str.title()

df = df.drop_duplicates(
    subset=["Customer", "Product", "Quantity", "UnitPrice", "City"],
    keep="first"
)

print("Exact duplicate rows remaining:", df.duplicated().sum())
```

```python
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["UnitPrice"] = pd.to_numeric(df["UnitPrice"], errors="coerce")

print("Invalid quantity values:", df["Quantity"].isna().sum())
print("Invalid price values:", df["UnitPrice"].isna().sum())

df = df.dropna(
    subset=["Quantity", "UnitPrice", "City", "Product", "Customer"]
).copy()

df = df[
    (df["Quantity"] > 0) &
    (df["UnitPrice"] > 0)
].copy()

print("Invalid prices remaining:", (df["UnitPrice"] <= 0).sum())
```

```python
highest_price_row = df.loc[df["UnitPrice"].idxmax()]

print("Highest-priced product:")
print("Product:", highest_price_row["Product"])
print("Unit price: ₹", highest_price_row["UnitPrice"])
```

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]

city_revenue = (
    df.groupby("City")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print("Revenue by city:")
print(city_revenue)

print("\nHighest-revenue city:", city_revenue.idxmax())
print("Revenue:", city_revenue.max())
```

```python
print("Final dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())

print("\nExact duplicate rows:", df.duplicated().sum())

print("\nInvalid prices:", (df["UnitPrice"] <= 0).sum())

print("Invalid quantities:", (df["Quantity"] <= 0).sum())

print("\nData types:")
print(df.dtypes)

print("\nFinal cleaned dataset:")
print(df)
```

```python
df.to_csv("cleaned_ecommerce.csv", index=False)

print("Cleaned dataset saved successfully!")
```
