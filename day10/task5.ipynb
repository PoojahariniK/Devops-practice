```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Zone': ['Central', 'North', 'South', 'East', 'West'] * 3,
    'Time': ['Morning'] * 5 + ['Evening'] * 5 + ['Night'] * 5,
    'Vehicles': [900, 500, 450, 650, 400,
                 1100, 700, 600, 800, 550,
                 300, 200, 180, 250, 150],
    'Speed': [20, 35, 40, 30, 45,
              12, 25, 32, 18, 38,
              45, 55, 58, 50, 60],
    'PM25': [80, 40, 30, 55, 25,
             100, 60, 45, 75, 35,
             25, 18, 15, 20, 12]
}

df = pd.DataFrame(data)
df
```

```python
vehicles_arr = np.array(df['Vehicles'])
pm25_arr = np.array(df['PM25'])

avg_vehicles = np.mean(vehicles_arr)
min_vehicles = np.min(vehicles_arr)
max_vehicles = np.max(vehicles_arr)
total_vehicles = np.sum(vehicles_arr)

print("Average:", avg_vehicles)
print("Minimum:", min_vehicles)
print("Maximum:", max_vehicles)
print("Total:", total_vehicles)
```

```python
df['Vehicles_in_Hundreds'] = df['Vehicles'] / 100
df['Speed_mps'] = df['Speed'] / 3.6
df['PM25_per_Vehicle'] = df['PM25'] / df['Vehicles']

df.head()
```

```python
# Average, minimum, maximum and total
df['Vehicles'].mean()
```

```python
# High pollution observations
high_pollution = df[df['PM25'] > 50]
high_pollution
```

```python
# High traffic and low speed
congestion = df[(df['Vehicles'] > 600) & (df['Speed'] < 25)]
congestion
```

```python
# Sort by pollution, highest first
df.sort_values(by='PM25', ascending=False)
```

```python
df.groupby('Time')[['Vehicles', 'Speed', 'PM25']].mean()
```

```python
correlation = df[['Vehicles', 'Speed', 'PM25']].corr()
correlation
```

```python
df['Predicted_PM25'] = df.groupby('Time')['PM25'].transform('mean')
df[['Zone', 'Time', 'PM25', 'Predicted_PM25']]
```

```python
df['Action'] = np.where(
    (df['Vehicles'] > 600) & (df['Speed'] < 25),
    'Investigate congestion',
    'Continue monitoring'
)

df[['Zone', 'Time', 'Vehicles', 'Speed', 'Action']]
```
